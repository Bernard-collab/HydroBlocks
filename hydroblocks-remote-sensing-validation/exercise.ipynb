```python
# create a time series data from 19100 to 2000; Time and a variable like rainfall or temperature; then plot the time series data using matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

Rainfall = np.random.rand(100) * 200   #100 random numbers between 0 and 200
Time = pd.date_range(start='1900-01-01', periods=100, freq='YE')
df = pd.DataFrame({'Time': Time, 'Rainfall': Rainfall})
df

# set a threshold for the variable (rainfall > 100 mm) and count the number of times the variable exceeds the threshold each year; 
threhold = 100

# then plot the time series data to check the distribution of the variable over time;

df['Exceeds Threshold'] = df['Rainfall'] > threhold
df['Exceeds Threshold'].value_counts()

plt.figure(figsize=(10, 5))
plt.plot(df['Time'], df['Rainfall'], label='Rainfall')
plt.axhline(y=threhold, color='r', linestyle='--', label=f'Threshold = {threhold} mm')
plt.xlabel('Time')
plt.ylabel('Rainfall (mm)')
plt.title('Time Series of Rainfall')
plt.legend()
plt.show()
```

```python
# check time range for the longest period of time where the variable exceeds the threshold;

x= df['Rainfall'] 
max_time= 0
for i in range(len(x)):
    if x[i] > threhold:
        max_time += 1
    else:
        max_time = 0
print("Longest period of time where rainfall exceeds the threshold:", max_time, "years")



    
```

```python
# Try
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
time = np.arange(100)
lat = np.linspace(-90, 90, 100)
lon = np.linspace(-180, 180, 250)

shape = (100, 100, 250) #data shape (time, lat, lon)
rainfall = np.random.exponential(scale=10, size=shape).astype("float32")

rain_xr = xr.DataArray(rainfall, coords={"time": time, "lat": lat, "lon": lon},dims=("time", "lat", "lon"),name="rainfall")

rain_xr = rain_xr.mean(dim=("lat","lon"))

plt.plot(rain_xr)
plt.ylabel("Mean Rainfall amount")
plt.xlabel("time index")
plt.title("Plot of rainfal for 100 timesteps")
```
